```python
print("PDM:")
print(data["PDM"].value_counts(dropna=False))

print("\nSDC:")
print(data["SDC"].value_counts(dropna=False))

print("\nVenta:")
print(data["Venta"].value_counts(dropna=False))
```

```python
# Reemplazar valores nulos de PDM, SDC y Venta utilizando la moda

data["PDM"] = data["PDM"].fillna(data["PDM"].mode()[0])

data["SDC"] = data["SDC"].fillna(data["SDC"].mode()[0])

data["Venta"] = data["Venta"].fillna(data["Venta"].mode()[0])
```

```python
print("Nulos en PDM:", data["PDM"].isnull().sum())
print("Nulos en SDC:", data["SDC"].isnull().sum())
print("Nulos en Venta:", data["Venta"].isnull().sum())
```

```python
# Eliminar espacios al inicio y al final de los nombres de las columnas
data.columns = data.columns.str.strip()
```

```python
print("BDC:")
print(data["BDC"].value_counts(dropna=False))

print("\nNombre Cliente:")
print(data["Nombre Cliente"].value_counts(dropna=False).head(10))

print("\nEstatus de Lead:")
print(data["Estatus de Lead"].value_counts(dropna=False))

print("\nAsesor Asignado:")
print(data["Asesor Asignado"].value_counts(dropna=False).head(20))
```

```python
# Tratamiento de valores nulos en variables categóricas

data["BDC"] = data["BDC"].fillna("Sin asignar")

data["Nombre Cliente"] = data["Nombre Cliente"].fillna("Sin información")

data["Estatus de Lead"] = data["Estatus de Lead"].fillna(
    data["Estatus de Lead"].mode()[0]
)

data["Asesor Asignado"] = data["Asesor Asignado"].fillna("Sin asignar")
```

```python
print("Nulos en BDC:", data["BDC"].isnull().sum())
print("Nulos en Nombre Cliente:", data["Nombre Cliente"].isnull().sum())
print("Nulos en Estatus de Lead:", data["Estatus de Lead"].isnull().sum())
print("Nulos en Asesor Asignado:", data["Asesor Asignado"].isnull().sum())
```

```python
print("Fecha:")
print(data["Fecha"].value_counts(dropna=False).head(15))

print("\nTelefono:")
print(data["Telefono"].value_counts(dropna=False))

print("\nPotencial de compra:")
print(data["Potencial de compra"].value_counts(dropna=False))

print("\nUnnamed: 15:")
print(data["Unnamed: 15"].value_counts(dropna=False).head(20))
```

```python
data = data.drop(columns=["Unnamed: 15"])
```

```python
data["Fecha"] = data["Fecha"].fillna("Sin fecha registrada")
data["Telefono"] = data["Telefono"].fillna("Sin información")
data["Potencial de compra"] = data["Potencial de compra"].fillna(
    data["Potencial de compra"].mode()[0]
)
```

```python
print("Nulos en Fecha:", data["Fecha"].isnull().sum())
print("Nulos en Telefono:", data["Telefono"].isnull().sum())
print("Nulos en Potencial de compra:", data["Potencial de compra"].isnull().sum())

print("\n¿Existe Unnamed: 15?")
print("Unnamed: 15" in data.columns)
```

```python
# Tratamiento de valores nulos en preguntas abiertas

data["¿Por que no ha visitado la agencia?"] = data["¿Por que no ha visitado la agencia?"].fillna("Sin respuesta")

data["¿Por que no hay solicitud de credito?"] = data["¿Por que no hay solicitud de credito?"].fillna("Sin respuesta")

data["¿Por que no hubo prueba de manejo ?"] = data["¿Por que no hubo prueba de manejo ?"].fillna("Sin respuesta")

data["¿Por que no hubo proceso wow?"] = data["¿Por que no hubo proceso wow?"].fillna("Sin respuesta")

data["¿Por que se asigno antes de vistar piso?"] = data["¿Por que se asigno antes de vistar piso?"].fillna("Sin respuesta")
```
